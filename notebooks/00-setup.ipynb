# 00-setup

## Purpose
Set up the environment, document data access, and define the analysis contract for the capstone.

## Environment
- Python 3
- duckdb
- pandas
- scikit-learn
- matplotlib
- jupyter / nbconvert

## Data access
- Use the FlyRank ML Internship dataset release from the assignment resources.
- Keep tokens and raw exports private.
- Document the exact release id and table names in the capstone notebook.

## Analysis contract
1. Build page-level and query-level features only from information available before the prediction window.
2. Use a chronological split to reduce leakage.
3. Compare the model to a naive baseline.
4. Convert scores into a ranked action list.

# Ranking Signal Analysis Capstone

## Title
Ranking Signal Analysis for FlyRank Content

## Abstract
This project asks which safe content and search signals are most associated with stronger visibility and engagement in FlyRank data. I built a repeatable ranking-signal pipeline that aggregates page-level features from the warehouse, constructs a forward-looking target, and compares a simple baseline with a tree-based model. The evaluation uses a time-aware split to reduce leakage and to better reflect how a ranking recommendation would be used in practice. The model is interpreted as an evidence-backed ranking engine rather than a causal proof of what the search system does. The output is a ranked action list that prioritizes pages most likely to benefit from content, metadata, or monitoring review.

## Introduction / Problem statement
The decision supported by this work is simple: which pages should content or SEO teams review first? This analysis turns historical page and query behavior into a ranked list of opportunities so teams can focus on the pages with the clearest improvement potential.

## Methodology
- Assumptions: historical page-level signals can rank pages for review; the model is predictive, not causal.
- Features: title length, content length, freshness, query count, click count, impression count, CTR, engagement rate, trend deltas, and recency-weighted aggregates.
- Label definition: a forward-window opportunity score, such as whether a page improves in clicks, engagement, or visibility over a future period.
- Baseline: a naive heuristic ranking such as recent clicks, recent CTR, or a linear score from a small feature set.
- Validation: time-aware split with leakage checks that prevent future-window features from entering training.

## Results
Compare the model against the baseline on the same holdout split. Report ranking metrics such as precision@k, hit rate@k, or lift over baseline, plus error analysis. Include charts showing model scores versus realized outcomes and score distributions by page group.

## Limitations & honest framing
This is an observational ranking model, not a causal experiment. The recommendations are decision-support signals, not proof that a change will improve search performance. Feature quality depends on warehouse completeness and the chosen release window.

## Ranked recommendations
1. Review high-impression / low-CTR pages first.
2. Review declining freshness-sensitive pages next.
3. Monitor pages with strong engagement but low visibility.
